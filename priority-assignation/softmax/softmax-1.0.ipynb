```python
# Functions

from typing import List

def systolic_blood_pressure(s):
    if 90 <= s <= 120:
        return 0
    if s < 90:
        return (s - 90) ** 2
    if 140 >= s > 120:
        return abs(s - 120) ** 1.2
    else:
        return 36.411 + ((s - 140) ** 2)

def diastolic_blood_pressure(s):
    if 60 <= s <= 80:
        return 0
    if s < 60:
        return (s - 60) ** 2
    if 90 >= s > 80:
        return abs(s - 80) ** 2
    else:
        return 100 + ((s - 90) ** 2)

def map(s):
    if 70 <= s <= 100:
        return 0
    if s < 70:
        return (s - 70) ** 2
    if 100 <= s < 110:
        return abs(s - 100)
    else:
        return 10 + (abs(s - 110) ** 2)

def pulse_pressure(p):
    return (p - 40) ** 2

def respiratory_rate(age: List[float], respiratory_rate: List[float]) -> List[float]:
    res = [0.0] * len(age)
    for i in range(len(age)):
        if age[i] < 1:
            res[i] = abs(respiratory_rate[i] - 45) ** 2
            continue
        elif age[i] < 2:
            res[i] = abs(respiratory_rate[i] - 30) ** 2
            continue
        elif 3 < age[i] < 6:
            res[i] = abs(respiratory_rate[i] - 25) ** 2
            continue
        elif 6 < age[i] < 12:
            res[i] = abs(respiratory_rate[i] - 22.5) ** 2
            continue
        else:
            res[i] = abs(respiratory_rate[i] - 16) ** 2
            continue
    return res

def temp(t):
    if 36 < t < 38:
        return 0
    if t > 38:
        return (t - 38) ** 2
    return abs(36 - t) ** 2

def bmi(b):
    if 18.5 <= b <= 25:
        return 0
    if b < 18.5:
        return (b - 18.5) ** 2
    else:
        return (25 - b) ** 2

# Modified functions

def systolic_blood_pressure_mod(s):
    if 90 <= s <= 120:
        return 0
    if s < 90:
        return (s - 90) ** 3
    if 140 >= s > 120:
        return abs(s - 120) ** 1.2
    else:
        return 36.411 + ((s - 140) ** 3)

def diastolic_blood_pressure_mod(s):
    if 60 <= s <= 80:
        return 0
    if s < 60:
        return (s - 60) ** 3
    if 90 >= s > 80:
        return abs(s - 80) ** 2
    else:
        return 100 + ((s - 90) ** 3)

def map_mod(s):
    if 70 <= s <= 100:
        return 0
    if s < 70:
        return (s - 70) ** 3
    if 100 <= s < 110:
        return abs(s - 100)
    else:
        return 10 + (abs(s - 110) ** 3)

def pulse_pressure_mod(p):
    return (p - 40) ** 3

def respiratory_rate_mod(age: List[float], respiratory_rate: List[float]) -> List[float]:
    res = [0.0] * len(age)
    for i in range(len(age)):
        if age[i] < 1:
            res[i] = abs(respiratory_rate[i] - 45) ** 3
            continue
        elif age[i] < 2:
            res[i] = abs(respiratory_rate[i] - 30) ** 3
            continue
        elif 3 < age[i] < 6:
            res[i] = abs(respiratory_rate[i] - 25) ** 3
            continue
        elif 6 < age[i] < 12:
            res[i] = abs(respiratory_rate[i] - 22.5) ** 3
            continue
        else:
            res[i] = abs(respiratory_rate[i] - 16) ** 3
            continue
    return res

def temp_mod(t):
    if 36 < t < 38:
        return 0
    if t > 38:
        return (t - 38) ** 3
    return abs(36 - t) ** 3

def bmi_mod(b):
    if 18.5 <= b <= 25:
        return 0
    if b < 18.5:
        return (b - 18.5) ** 3
    else:
        return (25 - b) ** 3
```

```python
from tensorflow.keras.layers import Dense
from tensorflow.keras.losses import SparseCategoricalCrossentropy
from tensorflow.keras.models import Sequential
from tensorflow.nn import softmax

model = Sequential([
    Dense(units=25, activation='relu'),
    Dense(units=15, activation='relu'),
    Dense(units=6, activation='linear')
])

model.compile(optimizer='adam', loss = SparseCategoricalCrossentropy(from_logits=True), metrics=['accuracy'])
model.fit(x_train, y_train, epochs=200)

logits = model(x_train)

f_x = softmax(logits)

f_x
```

```python
f_x = np.array(f_x)
f_x = f_x.argmax(axis=1)
f_x[:25]
```

```python
y = np.array(y_train)
y[:25]
```

```python
correct = np.count_nonzero(np.where(y == f_x, 1, 0))
print(correct)
```
